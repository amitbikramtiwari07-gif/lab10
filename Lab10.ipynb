EXP 10: K-Means Clustering

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn
from sklearn.cluster import KMeans
```

```python
sns.set()
```

```python
#Load the data
data=pd.read_csv("3.01.Countryclusters.csv")
print(data)
```

```python
#Plotting the dataset
plt.scatter(data["Latitude"],data["Longitude"])
plt.xlim(-180,180)
plt.ylim(-90,90)
plt.show()
```

```python
#Selecting the Feature
x=data.iloc[:,1:3]
```

```python
#Clustering
kmeans=KMeans(5)
kmeans.fit(x)
#Clustering Result
#fitting the data into model
identified_clusters=kmeans.fit_predict(x)
#identified clusters where our data can go
identified_clusters
#Creating a copy of our data
data_with_clusters=data.copy()
#new CLusters---> data
data_with_clusters['Clusters']=identified_clusters
#Plot the data
plt.scatter(data_with_clusters["Latitude"],data_with_clusters["Longitude"],c=data_with_clusters ["Clusters"],cmap='rainbow')
plt.show()
```
